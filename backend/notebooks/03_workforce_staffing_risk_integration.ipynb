## 1. Problem Definition & Objectives
Connecting attrition probability and leave demand directly to operational workforce availability:
$$\text{Available Workforce} = \text{Current Active} - \text{Predicted Leave} - \text{Predicted Attrition} - \text{Operational Buffer}$$
$$\text{Staffing Gap} = \text{Required Workers} - \text{Available Workforce}$$

```python
import pandas as pd
import numpy as np

hubs_data = [
    {'hub': 'Lucena Delivery Center', 'current_active': 58, 'expected_workload': 9240, 'required': 52, 'pred_leave': 4, 'pred_attrition': 1, 'buffer': 2},
    {'hub': 'Tayabas Logistics Hub', 'current_active': 34, 'expected_workload': 5600, 'required': 33, 'pred_leave': 3, 'pred_attrition': 1, 'buffer': 1},
    {'hub': 'Quezon Distribution Center', 'current_active': 22, 'expected_workload': 3900, 'required': 24, 'pred_leave': 2, 'pred_attrition': 1, 'buffer': 1},
]
df_risk = pd.DataFrame(hubs_data)
df_risk['estimated_available'] = df_risk['current_active'] - df_risk['pred_leave'] - df_risk['pred_attrition'] - df_risk['buffer']
df_risk['staffing_gap'] = df_risk['required'] - df_risk['estimated_available']

def classify_risk(row):
    gap = row['staffing_gap']
    req = row['required']
    ratio = gap / req
    if ratio >= 0.20: return 'Critical Risk'
    elif ratio >= 0.10: return 'High Risk'
    elif ratio > 0: return 'Medium Risk'
    return 'Low Risk'

df_risk['risk_level'] = df_risk.apply(classify_risk, axis=1)
print(df_risk[['hub', 'current_active', 'required', 'estimated_available', 'staffing_gap', 'risk_level']])
```
